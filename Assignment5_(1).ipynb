```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload a low-contrast image
uploaded = files.upload()

# Load first uploaded image
for fname in uploaded.keys():
    img = cv2.imread(fname, cv2.IMREAD_GRAYSCALE)
    break

plt.imshow(img, cmap='gray')
plt.title('Original Low-Contrast Image')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(img.ravel(), bins=256, color='gray')
plt.title('Histogram of Original Image')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.show()
```

```python
# Find minimum and maximum intensity values
min_val, max_val = np.min(img), np.max(img)

# Apply contrast stretching formula
stretched = ((img - min_val) / (max_val - min_val)) * 255
stretched = np.uint8(stretched)

# Display result
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(stretched, cmap='gray')
plt.title('Contrast Stretched')
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(img.ravel(), bins=256, color='gray')
plt.title('Original Histogram')

plt.subplot(1,2,2)
plt.hist(stretched.ravel(), bins=256, color='gray')
plt.title('Contrast Stretched Histogram')

plt.tight_layout()
plt.show()
```

```python
equalized = cv2.equalizeHist(img)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(equalized, cmap='gray')
plt.title('Histogram Equalized')
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.hist(img.ravel(), bins=256, color='gray')
plt.title('Original Histogram')

plt.subplot(1,2,2)
plt.hist(equalized.ravel(), bins=256, color='gray')
plt.title('Equalized Histogram')

plt.tight_layout()
plt.show()
```

```python
print("""
Observations:
1. The original image had a narrow intensity range — low contrast.
2. After contrast stretching, pixel values were redistributed across the full 0–255 range,
   making the image brighter and clearer.
3. Histogram Equalization further enhanced global contrast by making the pixel intensity
   distribution more uniform.
4. Equalization may sometimes over-enhance noise, while contrast stretching maintains smoother transitions.
""")
```
